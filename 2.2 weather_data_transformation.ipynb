```python
import pandas as pd
import json
import os
```

```python
def loadWeatherJson(path):
    with open(f"{path}", "r") as f:
        weather = json.load(f)

    return pd.DataFrame({
        "city": "Lahore",
        "date": weather["daily"]["time"],
        "max_temp": weather["daily"]["temperature_2m_max"],
        "rain_mm": weather["daily"]["precipitation_sum"]
    })
```

```python
lahore = loadWeatherJson("./raw/weather_lahore_2026-11-05.json")
lahore.head()
```

```python
islamabad = loadWeatherJson("./raw/weather_islamabad_2026-11-05.json")
islamabad.head()
```

```python
karachi = loadWeatherJson("./raw/weather_karachi_2026-11-05.json")
karachi.head()
```

```python
weather_df = pd.concat( [islamabad, lahore, karachi], ignore_index=True )
```
