```python
import pandas as pd
import os
```

```python
orders_df = pd.read_csv("./cleaned/cleaned_orders.csv")
```

```python
weather_df = pd.read_csv("./cleaned/weather.csv")
```

```python
customers_df = pd.read_csv("./raw/customers.csv")
```

```python
orders_df.info()
```

```python
customers_df.info()
```

```python
orders_df = orders_df.merge(
    customers_df[["customer_id", "city"]],
    left_on="customer_id",
    right_on="customer_id",
    how="left"
)
```

```python
orders_df
```

```python
orders_df["date"] = pd.to_datetime(orders_df["order_date"], format="mixed").dt.normalize()
weather_df["date"] = pd.to_datetime(weather_df["date"], format="mixed").dt.normalize()
```

```python
orders_df.head()
```

```python
weather_df.head()
```

```python
orders = orders_df.merge(
    weather_df,
    on=["city", "date"],
    how="left"
)
```

```python
orders
```

```python
os.makedirs("./final_data", exist_ok=True)
orders.to_csv("./final_data/orders.csv", index=False)
```
