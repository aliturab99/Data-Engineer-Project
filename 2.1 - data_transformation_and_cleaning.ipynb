```python
import pandas as pd
import json
```

```python
m_orders = pd.read_json('./raw/mongo_orders.json')
pg_orders = pd.read_csv("./raw/orders.csv")
```

```python
def setCustomerId(data):
    # print(data)
    if data['customer_id'] >= 0:
        return int(data['customer_id'])
    else:
        return int(data['customer']['id'])
```

```python
m_orders["customer_id"] = m_orders[["customer_id", "customer"]].apply(setCustomerId, axis=1)
```

```python
def setItemsData(data, name):
    return data[0][name]
```

```python
m_orders["category"] = m_orders["items"].apply(lambda x: setItemsData(x, "category"))
m_orders["quantity"] = m_orders["items"].apply(lambda x: setItemsData(x, "quantity"))
m_orders.drop(columns=["items"], inplace=True)
```

```python
def setPaymentMethod(data):
    return data["method"]
```

```python
m_orders["payment_method"] = m_orders["payment"].apply(lambda x: setPaymentMethod(x))
m_orders.drop(columns=["payment"], inplace=True)
```

```python
def getCityName(data):
    if data == None:
        return None
    else:
        return data['city']
```

```python
m_orders["city"] = m_orders["customer"].apply(lambda x: getCityName(x))
```

```python
m_orders["order_date"] = pd.to_datetime(m_orders["order_date"])
```
